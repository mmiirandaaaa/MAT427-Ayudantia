## Introducción

En esta ayudantía veremos qué es una base de Gröbner y algunas aplicaciones. Para ello partamos motivando con el siguiente problema.

El *problema de pertenencia a un ideal* es: Dado un polinomio $f\in k[x_1,\dots,x_n]$ y un ideal $I\subseteq k[x_1,\dots,x_n]$, determinar si $f$ pertenece al ideal $I$.

## Caso univariado

Sea $k$ un cuerpo algebraicamente cerrado.
Sabemos que $k[x]$ es un dominio euclideano cuya "función euclideana" es el grado del polinomio $deg(\cdot): k[x] \to \mathbb{N}$. Usando esta función podemos construir un algoritmo para la división de polinomios.

Dados $f,g\in k[x]$ queremos encontrar un cociente $q\in k[x]$ y un resto $r\in k[x]$ tales que $f=qg +r$ con $deg(r) = 0$ o $deg(r) < deg(g)$. El algoritmo que calcula el cociente y el resto de la división lo llamamos *algoritmo de división* y lo podemos describir mediante la siguiente secuencia de pasos.
- **Etapa 1:** Inicializa $F=f$ y $Q=0$.
- **Etapa 2:** Mientras $deg(F) > deg(q)$, repita:
$$
\begin{align*}
F &= F - \frac{lc(F)}{lc(g)}x^{n-m}g \\
Q &= Q + \frac{lc(F)}{lc(g)}x^{n-m}
\end{align*}
$$
donde $lc(f)$ es el coeficiente lider (*leading coefficient*) de $f$, $n := deg(F)$ y $m:=deg(g)$.
- **Etapa 3:** Retorna $resto = F$, $cociente = Q$.

Esto resulta útil para checkear pertenencia a un ideal. Dado un polinomio $f\in k[x]$ y un ideal $I=\langle g_1,\dots,g_s \rangle$ si acaso $f\in I$. Dado que $k[x]$ es un dominio euclideano, sabemos que podemos reducir los generadores de $I$ a uno solo, específicamente a $d := mcd(g_1,\dots,g_s)$. De esta manera sólo debemos dividir $f$ por $d$ y ver si el resto es cero o no.

```python
f = x**3 - x**2 + x - 1
g1 = x**4 - 2*x**3 + 4*x**2 + 2*x - 5
g2 = x**5 + x**3 - 2*x

print("Sean")
display(Math(f"f = {sympy.latex(f)}"))
display(Math(f"g_1 = {sympy.latex(f)}"))
display(Math(f"g_2 = {sympy.latex(f)}"))
print("Queremos checkear si se cumple")
display(Math(f"f \\in I := \\langle g_1, g_2 \\rangle"))
print("Para esto vemos que")
display(Math("d = x^2 - 1"))
print("Luego basta checkear el resto de dividir f por d")
q,r = sympy.div(f,x**2 - 1, domain='QQ')
display(Math(f"f = ({sympy.latex(q)})d + {sympy.latex(r)}"))
```

## Caso multivariable

Al aumentar la cantidad de variables que consideramos perdemos una propiedad valiosa de $k[x_1]$, ser un dominio euclideano (nos da el algoritmo de división) y, por lo tanto, ser un DIP.

Entonces queremos llevar el concepto de *grado* de un polinomio en una variable a uno en varias variables. La manera en la que haremos esto será mediante la definición de **órdenes monomiales**.

### Órdenes monomiales

Sea $\mathbb{T}^n$ el conjunto de monomios en variables $x_1,\dots,x_n$ junto con la multiplicación (monoide conmutativo).

Notemos que para $n=1$, la función 
$$
\begin{align*}
deg: \mathbb{T} &\to \mathbb{N} \\
x^{k} &\mapsto k
\end{align*}
$$
induce un orden total en $\mathbb{T}$. Además, las propiedades
- $deg(x^{k}) = 1$ si, y sólo si, $k=0$.
- Si $deg(x^{k}) > deg(x^{k'})$, entonces $deg(x^{l})deg(x^{k}) > deg(x^{l})deg(x^{k'})$.

Aseguran que cualquier subconjunto de $\mathbb{T}$ tiene un elemento mínimo, es decir, el orden inducido por la función $deg$ es un buen orden en $\mathbb{T}$.

Estas propiedades son exactamente las que queremos imponer a un orden total en $\mathbb{T}^n$. De manera más formal, diremos que un orden total $>$ en $\mathbb{T}^n$ es un *órden monomial* si cumple las siguientes propiedades:
- $x^{\alpha} > 1$ siempre que $\alpha\not=0$.
- Si $\mu > \mu'$, entonces $\nu\mu > \nu\mu'$ para cualquier $\nu,\mu,\mu'\in\mathbb{T}^n$.

**Proposición 1.** Todo órden monomial es un buen orden.

*Demostración.* Suponga, por contradicción que un orden monomial $>$ en $\mathbb{T}^n$ no es un buen orden. Entonces existe una cadena decreciente infinita, llamemosla $S$. Sea $I\subseteq k[x_1,\dots,x_n]$ el ideal generado por $S$. Usando los *Lemas 1 y 2* sabemos que existe un subconjunto finito de monomios $G\subseteq S$ que genera $I$. Como $S$ es una secuencia decreciente y $G$ es finito, existe un $\mu\in S$ cota inferior de $G$. Por otro lado, como $\mu\in I$ es un monomio, debe existir $\nu\in G$ tal que $\mu=\sigma\nu$ para algún $\sigma\in\mathbb{T}^n\setminus\{1\}$. Luego
$$
\nu > \mu = \sigma\nu > \nu
$$
Contradicción.

**Ejemplos de órdenes monomiales son: lex, glex y grevlex.**

#### Algunos resultados útiles
**Lema 1.** Sea $I$ un ideal finitamente generado. Si $S$ es un conjunto infinito generador de $I$, entonces existe un subconjunto finito de $S$ que también genera $I$.

*Demostración* Sea $I = \langle g_1,\dots,g_l\rangle$. Como $S$ también genera $I$, cada $g_i$ es una combinación lineal finita de elementos de $S$. Entonces $g_j \in \langle s_1, \dots, s_{r(j)}\rangle$, escogiendo $m:=\max_{j}r(j)$ tenemos que $\langle g_1,\dots,g_l\rangle \subseteq \langle s_1, \dots, s_m\rangle \subseteq I$.

**Lema 2.** Todo ideal monomial en $k[x_1,\dots,x_n]$ es finitamente generado.

*Demostración* Sea I un ideal monomial. Suponga que no es finitamente generado, entonces podemos armar la siguiente secuencia
$$
\begin{align*}
\mu_1 &:= \min_{\mu\in I} deg_{x_n}(\mu) \\
\mu_2 &:= \min_{\mu\in I\setminus\langle\mu_1\rangle} deg_{x_n}(\mu) \\
&\vdots \\
\mu_i &:= \min_{\mu\in I\setminus\langle\mu_1,\dots,\mu_{i-1}\rangle} deg_{x_n}(\mu) \\
&\vdots \\
\end{align*}
$$
Pero $\mu_i = \nu_i x_n^{k_i}$ con $\nu_i\in k[x_1,\dots,x_{n-1}]$. Veamos que $J:=\langle \nu_i\mid i\in\mathbb{N}\rangle$ no puede ser finitamente generado.

Supongamos que sí lo es, entonces por el *Lema 1* tenemos que $J=\langle v_1,\dots, v_m\rangle$ para algún entero $m>0$. Entonces $\nu_{m+1} = \nu_l\eta$ para algún $l\in\{1,\dots,m\}$ y algún monomio $\eta\in k[x_1,\dots,x_{n-1}]$. Luego
$$
\mu_{m+1}=\nu_{m+1}x_n^{k_m} = \eta x_n^{k_m - k_l}\mu_l
$$
lo cual implica que $\mu_{m+1}\in\langle\mu_1,\dots,\mu_m\rangle$. Contradicción con la manera en que $\mu_{m+1}$ fue escogido.

### División

Ahora podemos proponer el algoritmo de división en $k[x_1,\dots,x_n]$.

Un primer intento sería imitar el algoritmo que teníamos para $n=1$.

**Algoritmo de división euclideana** Recibe $f,g\in k[x]$ y retorna los únicos $q,r\in k[x]$ tales que $f=qg + r$ con $r=0$ o $r < g$.
- **Etapa 1:** Inicializa $F=f$ y $Q=0$.
- **Etapa 2:** Mientras $deg(F) > deg(q)$, repita:
$$
\begin{align*}
F &= F - \frac{lc(F)}{lc(g)}x^{n-m}g \\
Q &= Q + \frac{lc(F)}{lc(g)}x^{n-m}
\end{align*}
$$
- **Etapa 3:** Retorna $resto = F$, $cociente = Q$.

Pero esto nos trae problemas. Considere $f=x_1$ y $g=x_2$.

En nuestro segundo intento solucionamos ese problema simplemente declarando que, cuando no se pueda realizar la división de los térimnos líderes, restamos a $F$ su término líder, lo sumamos al resto, y pasamos a la siguiente iteración.

Por ejemplo

Dividir $f = 6x_1^6x_2^4 + 7x_1^5 + 8x_1^4x_2^7 + 1$ por $g = x_1^4x_2^2 + 1$. Obtenemos $Q=6x_1^2x_2^2 + 8x_2^5$ y $R=7x_1^5-6x_1^2x_2^2 - 8x_2^5 + 1$. Notamos que $R >_{lex} Q$. No es tal como en una variable.

La tercera es la vencida. Podemos mejorar un poco esto agregando múltiples divisores y dividiendo por el primero que podamos, entonces sólo pasamos el término líder de $F$ cuando para todos los términos líder de los divisores no se pudo efectuar la división. Esto nos conduce al siguiente algoritmo

**Algoritmo de división en varias variables** Recibe $f,g_1,\dots,g_s\in k[x_1,\dots,x_n]$ y un órden monomial $>$. Retorna cocientes $q_1,\dots,q_s\in k[x_1,\dots,x_n]$ y un resto $r\in k[x_1,\dots,x_n]$ tales que
- $f=q_1g_1 + \dots + q_sg_s + r$ y,
- $r=0$ o ningún monomio de $r$ es divisible por los términos $in(g_1), \dots, in(g_s)$.

- **Etapa 1:** Inicializa $F=f$ y $Q_1=\dots=Q_s=0$ y $R=0$.
- **Etapa 2:** Mientras $F \not= 0$, verificar si existe algún entero $i\in\{1,\dots,s\}$ para el cual $in(g_i)$ divide a $in(F)$, si existe escoja el menor índice y haga:
$$
\begin{align*}
F &= F - \frac{in(F)}{in(g_i)}g_i \\
Q_i &= Q_i + \frac{in(F)}{in(g_i)}
\end{align*}
$$
si no existe dicho índice $i$, entonces
$$
\begin{align*}
F &= F - in(F) \\
R &= R + in(F)
\end{align*}
$$
- **Etapa 3:** Retorna resto $= R$ y cocientes = $Q_1,\dots,Q_n$.

El algoritmo funciona pues cada vez reduce el lead term de $F$ y termina pues tenemos un buen orden.

Ejemplo con $f = 6x_1^6x_2^4 + 7x_1^5 + 8x_1^4x_2^7 + 1$ por $g_1 = x_1^4x_2^2 + 1$ y $g_2 = 2x_2^5 - x_2$.

O mejor $f = 2x_1x_4 - 2x_2x_3 - x_4 + x_2$ verificar si está en el ideal generado por $g_1 = x_1x_4 - x_2x_3 + x_4$ y $g_2 = x_1x_4 - x_3x_2 + x_2 - 2x_4$.
