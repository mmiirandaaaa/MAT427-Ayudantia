**Queremos dividir en varias variables (por qué?): Checkear pertenencia a un ideal**

Sea $k$ un cuerpo algebraicamente cerrado.
Sabemos que $k[x]$ es un dominio euclideano cuya "función euclideana" es el grado del polinomio $deg(\cdot): k[x] \to \mathbb{N}$. Usando esta función podemos construir un algoritmo para la división de polinomios.

Dados $f,g\in k[x]$ queremos encontrar un cociente $q\in k[x]$ y un resto $r\in k[x]$ tales que $f=qg +r$ con $deg(r) = 0$ o $deg(r) < deg(g)$. El algoritmo que calcula el cociente y el resto de la división lo llamamos *algoritmo de división* y lo podemos describir mediante la siguiente secuencia de pasos.
- **Etapa 1:** Inicializa $F=f$ y $Q=0$.
- **Etapa 2:** Mientras $deg(F) > deg(q)$, repita:
$$
\begin{align*}
F &= F - \frac{lc(F)}{lc(g)}x^{n-m}g \\
Q &= Q + \frac{lc(F)}{lc(g)}x^{n-m}
\end{align*}
$$
donde $lc(f)$ es el coeficiente lider (*leading coefficient*) de $f$, $n := deg(F)$ y $m:=deg(g)$.
- **Etapa 3:** Retorna $resto = F$, $cociente = Q$.

Esto resulta útil para checkear pertenencia a un ideal. Dado un polinomio $f\in k[x]$ y un ideal $I=\langle g_1,\dots,g_s \rangle$ si acaso $f\in I$. Dado que $k[x]$ es un dominio euclideano, sabemos que podemos reducir los generadores de $I$ a uno solo, específicamente a $d := mcd(g_1,\dots,g_s)$. De esta manera sólo debemos dividir $f$ por $d$ y ver si el resto es cero o no.

```python
f = x**3 - x**2 + x - 1
g1 = x**4 - 2*x**3 + 4*x**2 + 2*x - 5
g2 = x**5 + x**3 - 2*x

print("Queremos checkear si se cumple")
s = f"{sympy.latex(f)} \\in I := \\langle {sympy.latex(g1)}, {sympy.latex(g2)} \\rangle"
display(Math(s))
```

```python
# Calculando el mcd con el algoritmo euclideano
q, r = sympy.div(g2,g1, domain='QQ')
```
